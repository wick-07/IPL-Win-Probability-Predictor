```python
batsman_stat=(delivery_df.groupby(['batsman']).agg(
    runs=('batsman_runs','sum'),balls_faced=('ball','count')
    ).assign(batsman_sr=lambda x:100*x.runs/x.balls_faced).reset_index())
```

```python
batsman_stat
```

```python
bowlers_stat=(
    delivery_df.groupby('bowler').agg(
    run_conceded=('total_runs','sum'),
    balls_bowled=('ball','count'),
    wickets_took=('player_dismissed',lambda x:x.notna().sum())
).assign(
        economy=lambda x:6*x.run_conceded/x.balls_bowled,
        wicket_rate=lambda x:6*x.wickets_took/x.balls_bowled
        ).reset_index()
)
```

```python
bowlers_stat
```

```python
delivery_df.shape
```

```python
delivery_df.columns
```

```python
delivery_df=delivery_df.merge(batsman_stat[['batsman','batsman_sr']],on='batsman',how='left')
delivery_df=delivery_df.merge(bowlers_stat[['bowler','economy','wicket_rate']],on='bowler',how='left')
```

```python
delivery_df.shape
```

```python
categorical_cols = [
    'toss_winner', 'toss_decision', 'venue', 'batting_team', 'bowling_team'
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
preprocessing=ColumnTransformer([
    ('transform',
     OneHotEncoder(sparse_output=False,drop='first',handle_unknown='ignore'),
     categorical_cols)],remainder='passthrough')
```

```python
preprocessing
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
```

```python
pipe=Pipeline(
       steps=[('preprocess',preprocessing),
             ('model',LogisticRegression(solver='liblinear'))]
             )
```

```python
pipe.fit(X_train,Y_train)
```

```python
prediction=pipe.predict(X_test)
```

```python
prediction
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(Y_test,prediction)
```
